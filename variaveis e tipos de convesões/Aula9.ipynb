```python
# 1. Criando as variáveis (As caixinhas com etiquetas)
nome_piloto = "Luke"          # Tipo str (Texto)
nave = "X-Wing"               # Tipo str (Texto)
quantidade_naves = 10         # Tipo int (Inteiro)
velocidade_hiperespaco = 4.5  # Tipo float (Decimal)
forca_ativa = True            # Tipo bool (Booleano)

# 2. O problema: Python não soma texto com número diretamente
# Se tentássemos fazer: total = "150000" + 5 -> O sistema quebraria!

# 3. Solução: Usando o Casting (A Força) para converter os dados com segurança
creditos_texto = "150000"
bonus_missao = 5

# Convertendo o texto em número inteiro para poder somar
creditos_finais = int(creditos_texto) + bonus_missao

# 4. Exibindo os resultados no console
print("--- RELATÓRIO DA MISSÃO JEDI ---")
print(f"Piloto: {nome_piloto} | Nave: {nave}")
print(f"Aliança possui {quantidade_naves} naves prontas.")
print(f"Velocidade atual: {velocidade_hiperespaco} anos-luz.")
print(f"Status da Força: {'Ativa' if forca_ativa else 'Inativa'}")
print(f"Total de créditos calculados com segurança: {creditos_finais}")

# 5. Demonstração de erro controlado (O Lado Sombrio)
try:
    # Isso vai gerar um ValueError porque "X-Wing" tem letras e não pode virar número
    int(nave)
except ValueError as e:
    print("\n[ALERTA DE ERRO] Cuidado com o Lado Sombrio: Letras não viram números!")
```
